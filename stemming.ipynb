#### Porter Stemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemmimg=PorterStemmer()
```

```python
words=['eat','eating','eaten','ate','push','pushed','history','congralutions']
```

```python
for word in words:
    print(word +"---->"+stemmimg.stem(word))
```

#### Regexp Stemmer

```python
from nltk.stem import RegexpStemmer
```

```python
regex=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
regex.stem('pushing')
```

```python
regex.stem('ingpushing')
```

```python
regex=RegexpStemmer('$ing|s$|e$|able$', min=4)
```

```python
regex.stem('ingpushing')
```

```python
regex=RegexpStemmer('ing|s$|e$|able$|', min=4) ## min = The minimum length of string to stem
```

```python
regex.stem('ingpushing')
```

#### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snow=SnowballStemmer('english') ### we have to choose a language  make sure that language wriiten in small
## letter
```

```python
words=['write','writes','written','history','eat','eaten','eating','congralution']
```

```python
for word in words :
    print(word+'---->'+snow.stem(word))
```

```python
### Differnce between porter and snow ball
print(f" porter stemming = {stemmimg.stem("fairly"),stemmimg.stem("sportingly")}")
print(f" snowball stemming = {snow.stem("fairly"),snow.stem("sportingly")}")
```
