```python
import pandas as pd 
import numpy as np 
import nltk
```

```python
df=pd.read_csv('SMSSpamCollection.txt',sep='\t',names=['labels','messages'])
```

```python
df
```

```python
df.shape
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
import re
lemmatizer=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(0,len(df)):
    message=re.sub('[^a-zA-z]',' ',df['messages'][i])
    message=message.lower()
    message=message.split()
    message=[lemmatizer.lemmatize(word,pos='v') for word in message if not word in  stopwords.words('english') ]
    message=' '.join(message)
    corpus.append(message)
```

```python
y=df['labels']
```

```python
y
```

```python
corpus
```

```python
y=pd.get_dummies(df['labels'])
```

## train and test split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.20,random_state=42)
```

# feature engineering


```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=2000,ngram_range=(1,2))
```

```python
X_train=cv.fit_transform(X_train)
```

```python
X_test=cv.transform(X_test)
```

## model training

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
NB=MultinomialNB()
```

```python
model=NB.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
