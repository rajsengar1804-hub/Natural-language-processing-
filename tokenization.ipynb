```python
from nltk.tokenize import sent_tokenize
```

```python

corpus="""Hello My name is raj pratap singh sengar.
i am studying in AUMP! Madhya Gwalior .
Ex Airmen in Indian Airforce .
"""
```

```python
sent_tokenize(corpus)
```

```python
documents=sent_tokenize(corpus)
```

```python
documents
```

```python
corpus2="""welcome to raj pratap singh sengar's github repo .
currently studying in AUMP ! MAdhya pradesh India.
Mahindra 's Scorpiq N 

"""
```

```python
documents2=sent_tokenize(corpus2)
```

```python
documents2
```

```python
## pargraph to word 
from nltk.tokenize import word_tokenize
```

```python
words=word_tokenize(corpus2)
```

```python
words
```

```python
## sentence to words
for sentence in documents2 :
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize ## punctation ko bhi separate words ke treat krrta hain 
```

```python
for sentence in documents2:
    print(wordpunct_tokenize(sentence))
```

```python
wordpunct_tokenize(corpus2)
```
