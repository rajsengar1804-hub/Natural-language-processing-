```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
import nltk
nltk.download('wordnet')
```

```python
words=['write','writes','written','history','eat','eaten','eating','congralution']
```

```python
'''
pos =part to speech 
pos=v verb 
pos=n noun
pos=a adjective 
pos=r adverb
'''
```

```python
for word in words :
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
from nltk.stem import SnowballStemmer
from nltk.stem import PorterStemmer
stemmimg=PorterStemmer()
snow=SnowballStemmer('english')
```

```python
print(f" porter stemming = {stemmimg.stem("fairly"),stemmimg.stem("sportingly")}")
print(f" snowball stemming = {snow.stem("fairly"),snow.stem("sportingly")}")
print(f" lemmizater = {lemmatizer.lemmatize("fairly",pos='a'),lemmatizer.lemmatize("sportingly",pos='a')}")
```
