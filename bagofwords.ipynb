```python
import pandas as pd 
import numpy as np 
import nltk 
import re 
nltk.download('stopwords')
```

```python
df=pd.read_csv('SMSSpamCollection.txt',sep='\t',names=['labels','messages'])
```

```python
df
```

```python
df.shape
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer,WordNetLemmatizer
```

```python
stemming=PorterStemmer()
```

```python
lemmatizer=WordNetLemmatizer()
```

#### DATA PREPROCESSING

```python
corpus=[]
for i in range(0,len(df)):
    message=re.sub('[a-zA-z]',' ',df['messages'][i])# for removing the special character
    message=message.lower()
    message=message.split()
    words=[lemmatizer.lemmatize(word,pos='v') for word in message if not word in  stopwords.words('english') ]
    message=' '.join(words)
    corpus.append(message)

    
```

```python
corpus
```

```python
corpus=[]
for i in range(0,len(df)):
    message=re.sub('[^a-zA-z]',' ',df['messages'][i])# for removing the special character
    message=message.lower()
    message=message.split()
    message=[lemmatizer.lemmatize(word,pos='v') for word in message if not word in  stopwords.words('english') ]
    message=' '.join(message)
    corpus.append(message)
```

```python
corpus
```

```python
corpus2=[]
for i in range(0,len(df)):
    review=re.sub('[^a-zA-z]',' ',df['messages'][i])
    review=review.lower()
    review=review.split()
    review=[stemming.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus2.append(review)
```
